```python
import os
import json

# Path to your annotations JSON file
annotations_path = "/home/sabdelazim/Desktop/Kaitlyn_Catalyst/ct_classifier/datasets/CaltechCT/eccv_18_annotation_files/train_annotations.json"

# Load the JSON file
with open(annotations_path, "r") as f:
    data = json.load(f)
    
# Extract all names from the categories
all_names = [cat["name"] for cat in data["categories"]]
print("All names:", all_names)

# Define the output text file path
output_txt = "/home/sabdelazim/Desktop/Kaitlyn_Catalyst/ct_classifier/datasets/CaltechCT/target_species.txt"

# Write each name to the output text file, one per line
with open(output_txt, "w") as out_file:
    for name in all_names:
        out_file.write(name + "\n")

print(f"Saved target species to {output_txt}")
```
